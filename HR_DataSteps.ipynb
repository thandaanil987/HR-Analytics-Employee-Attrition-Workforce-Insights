```python
import pandas as pd
import numpy as np

df = pd.read_csv("HR-Employee-Attrition.csv")

print("Dataset Shape:", df.shape)
```

```python

df.head()
```

```python
df.info()
```

```python

# Missing values
print(df.isnull().sum())
```

```python

# Duplicate rows
print("Duplicates:", df.duplicated().sum())
```

```python

# Unique values
for col in df.select_dtypes(include="object").columns:
    print(col, df[col].unique())
```

```python

# Check constant columns
for col in df.columns:
    if df[col].nunique() == 1:
        print("Constant column:", col)
```

```python

# Remove verified redundant columns
df = df.drop(
    columns=["EmployeeCount", "Over18", "StandardHours"],
    errors="ignore"
)
```

```python

# Validate employee ID
print("Duplicate Employee IDs:", df["EmployeeNumber"].duplicated().sum())
```

```python

# Export cleaned dataset
df.to_csv("hr_employee_cleaned.csv", index=False)
```
